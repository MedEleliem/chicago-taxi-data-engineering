## 2. RÃ¨gles mÃ©tier principales

### Identification

* `trip_id` doit Ãªtre renseignÃ©.
* `trip_id` est la clÃ© mÃ©tier principale du trajet.
* Les doublons doivent Ãªtre analysÃ©s.
* Plusieurs trajets peuvent avoir le mÃªme `taxi_id`.

### Dates et durÃ©e

* `trip_start_timestamp` doit Ãªtre valide.
* `trip_end_timestamp` doit Ãªtre valide.
* `trip_end_timestamp >= trip_start_timestamp`.
* `trip_seconds > 0`.
* La durÃ©e dÃ©clarÃ©e doit Ãªtre cohÃ©rente avec la diffÃ©rence entre les deux timestamps.

### Distance

* `trip_miles >= 0`.
* Une distance Ã©gale Ã  `0` est autorisÃ©e mais doit Ãªtre surveillÃ©e.
* Les distances trÃ¨s Ã©levÃ©es sont considÃ©rÃ©es comme des anomalies potentielles.

### Montants

* `fare >= 0`.
* `tips >= 0`.
* `tolls >= 0`.
* `extras >= 0`.
* `trip_total >= 0`.
* `trip_total` doit Ãªtre globalement cohÃ©rent avec les diffÃ©rents composants tarifaires.

### Paiement et compagnie

* `payment_type` doit Ãªtre standardisÃ©.
* Une nouvelle valeur de `payment_type` nâ€™est pas automatiquement une erreur.
* `company` doit Ãªtre standardisÃ©.
* Une nouvelle compagnie nâ€™est pas automatiquement une erreur.

### GÃ©ographie

* Les champs gÃ©ographiques peuvent Ãªtre nuls.
* Un champ gÃ©ographique manquant ne provoque pas automatiquement le rejet du trajet.
* Si une latitude existe : `-90 <= latitude <= 90`.
* Si une longitude existe : `-180 <= longitude <= 180`.
* Latitude et longitude doivent Ãªtre cohÃ©rentes entre elles.

````python
import pandas as pd
import numpy as np

import json
import hashlib
import logging
import uuid

from pathlib import Path
from datetime import datetime, timezone

# ============================================================
# COMMON DATAOPS UTILITIES
# ============================================================

PROJECT_ROOT = (
    Path.cwd().parent
    if Path.cwd().name == "notebooks"
    else Path.cwd()
)

BASE_DIR = PROJECT_ROOT / "data"

BRONZE_DIR = BASE_DIR / "bronze" / "chicago_taxi"
SILVER_DIR = BASE_DIR / "silver" / "chicago_taxi"

REPORT_DIR = BASE_DIR / "reports"
LOG_DIR = BASE_DIR / "logs"

for path in [
    BRONZE_DIR,
    SILVER_DIR,
    REPORT_DIR / "bronze",
    REPORT_DIR / "silver",
    LOG_DIR,
]:
    path.mkdir(parents=True, exist_ok=True)


def utc_now():
    return datetime.now(timezone.utc).isoformat()


def generate_run_id(layer):
    timestamp = datetime.now(timezone.utc).strftime("%Y%m%dT%H%M%SZ")
    short_uuid = uuid.uuid4().hex[:8]
    return f"{layer.upper()}_{timestamp}_{short_uuid}"


def sha256_file(path):
    sha = hashlib.sha256()

    with open(path, "rb") as f:
        for chunk in iter(lambda: f.read(1024 * 1024), b""):
            sha.update(chunk)

    return sha.hexdigest()


def save_json(data, path):
    path = Path(path)
    path.parent.mkdir(parents=True, exist_ok=True)

    with open(path, "w", encoding="utf-8") as f:
        json.dump(
            data,
            f,
            indent=2,
            ensure_ascii=False,
            default=str
        )


def build_column_profile(df):

    def safe_nunique(series):
        try:
            return series.nunique(dropna=True)
        except TypeError:
            return series.apply(
                lambda value:
                tuple(value)
                if isinstance(value, list)
                else value
            ).nunique(dropna=True)

    profile = pd.DataFrame({
        "column": df.columns,
        "dtype": df.dtypes.astype(str).values,
        "rows": len(df),
        "null_count": [df[c].isna().sum() for c in df.columns],
        "null_rate": [df[c].isna().mean() for c in df.columns],
        "unique_values": [
            safe_nunique(df[c])
            for c in df.columns
        ]
    })

    return profile


def create_logger(name, run_id):
    logger = logging.getLogger(f"{name}_{run_id}")
    logger.setLevel(logging.INFO)
    logger.propagate = False

    # avoid duplicate handlers in Jupyter
    if logger.handlers:
        logger.handlers.clear()

    formatter = logging.Formatter(
        "%(asctime)s | %(levelname)s | %(message)s"
    )

    file_handler = logging.FileHandler(
        LOG_DIR / f"{run_id}.log",
        encoding="utf-8"
    )
    file_handler.setFormatter(formatter)

    console_handler = logging.StreamHandler()
    console_handler.setFormatter(formatter)

    logger.addHandler(file_handler)
    logger.addHandler(console_handler)

    return logger
````

````python
def build_bronze(
    source_df,
    start_date=None,
    end_date=None,
    source_name="Chicago SODA API",
    run_id=None,
):
    # ========================================================
    # INITIALIZATION
    # ========================================================

    run_id = run_id or generate_run_id("BRZ")

    logger = create_logger("bronze",run_id)

    started_at = utc_now()

    output_dir = (
        BRONZE_DIR
        / f"run_id={run_id}"
    )

    output_dir.mkdir(
        parents=True,
        exist_ok=True
    )

    report_dir = (
        REPORT_DIR
        / "bronze"
        / run_id
    )

    report_dir.mkdir(
        parents=True,
        exist_ok=True
    )

    logger.info(
        f"Bronze run started | "
        f"run_id={run_id} | "
        f"rows={len(source_df)}"
    )

    manifest = {
        "run_id": run_id,
        "layer": "bronze",
        "dataset": "chicago_taxi",
        "source": source_name,

        "start_date": start_date,
        "end_date": end_date,

        "started_at": started_at,
        "finished_at": None,

        "status": "RUNNING",

        "input_rows": len(source_df),
        "output_rows": None,

        "output_path": None,

        "sha256": None,
        "error": None
    }

    save_json(
        manifest,
        report_dir / "manifest.json"
    )

    try:

        # ====================================================
        # BRONZE DATA
        # ====================================================

        bronze_df = source_df.copy()

        output_file = (
            output_dir
            / "data.csv"
        )

        bronze_df.to_csv(
            output_file,
            index=False
        )

        # ====================================================
        # COLUMN PROFILE
        # ====================================================

        column_profile = build_column_profile(
            bronze_df
        )

        column_profile.to_csv(
            report_dir / "column_profile.csv",
            index=False
        )

        # ====================================================
        # CHECKSUM
        # ====================================================

        file_hash = sha256_file(
            output_file
        )

        # ====================================================
        # FINAL MANIFEST
        # ====================================================

        manifest.update({
            "status": "SUCCESS",

            "finished_at": utc_now(),

            "output_rows": len(bronze_df),

            "output_path": str(output_file),

            "sha256": file_hash,

            "columns": bronze_df.columns.tolist(),

            "column_count": len(bronze_df.columns),
        })

        save_json(
            manifest,
            report_dir / "manifest.json"
        )

        logger.info(
            f"Bronze completed | "
            f"run_id={run_id} | "
            f"rows={len(bronze_df)} | "
            f"output={output_file}"
        )

        return (
            bronze_df,
            manifest
        )

    except Exception as exc:

        manifest.update({
            "status": "FAILED",
            "finished_at": utc_now(),
            "error": str(exc)
        })

        save_json(
            manifest,
            report_dir / "manifest.json"
        )

        logger.exception(
            f"Bronze failed | "
            f"run_id={run_id}"
        )

        raise
````

````python
def build_silver(
    bronze_df,
    bronze_run_id,
    run_id=None
):

    # ========================================================
    # INITIALIZATION
    # ========================================================

    run_id = run_id or generate_run_id(
        "SLV"
    )

    logger = create_logger(
        "silver",
        run_id
    )

    started_at = utc_now()

    output_dir = (
        SILVER_DIR
        / f"run_id={run_id}"
    )

    output_dir.mkdir(
        parents=True,
        exist_ok=True
    )

    report_dir = (
        REPORT_DIR
        / "silver"
        / run_id
    )

    report_dir.mkdir(
        parents=True,
        exist_ok=True
    )

    logger.info(
        f"Silver run started | "
        f"run_id={run_id} | "
        f"source_run_id={bronze_run_id} | "
        f"rows={len(bronze_df)}"
    )

    manifest = {

        "run_id": run_id,

        "source_run_id": bronze_run_id,

        "layer": "silver",

        "dataset": "chicago_taxi",

        "status": "RUNNING",

        "started_at": started_at,

        "finished_at": None,

        "input_rows": len(bronze_df),

        "valid_rows": None,

        "rejected_rows": None,

        "warning_rows": None,

        "error": None,
    }

    save_json(
        manifest,
        report_dir / "manifest.json"
    )

    try:

        silver = bronze_df.copy()

        # ====================================================
        # 1. COLUMN NORMALIZATION
        # ====================================================

        silver.columns = (
            silver.columns
            .str.strip()
            .str.lower()
            .str.replace(
                r"\s+",
                "_",
                regex=True
            )
        )

        # ====================================================
        # 2. TYPE CASTING
        # ====================================================

        date_cols = [
            "trip_start_timestamp",
            "trip_end_timestamp",
        ]

        for col in date_cols:

            silver[col] = pd.to_datetime(
                silver[col],
                errors="coerce"
            )

        numeric_cols = [

            "trip_seconds",
            "trip_miles",

            "fare",
            "tips",
            "tolls",
            "extras",
            "trip_total",

            "pickup_centroid_latitude",
            "pickup_centroid_longitude",

            "dropoff_centroid_latitude",
            "dropoff_centroid_longitude",
        ]

        for col in numeric_cols:

            silver[col] = pd.to_numeric(
                silver[col],
                errors="coerce"
            )

        for col in [

            "pickup_community_area",
            "dropoff_community_area",

        ]:

            silver[col] = (
                pd.to_numeric(
                    silver[col],
                    errors="coerce"
                )
                .astype("Int64")
            )

        # ====================================================
        # 3. ERROR / WARNING ARRAYS
        # ====================================================

        silver["quality_errors"] = [
            []
            for _ in range(len(silver))
        ]

        silver["quality_warnings"] = [
            []
            for _ in range(len(silver))
        ]

        def add_error(mask, rule):

            silver.loc[
                mask,
                "quality_errors"
            ] = (
                silver.loc[
                    mask,
                    "quality_errors"
                ]
                .apply(
                    lambda x:
                    x + [rule]
                )
            )

        def add_warning(mask, rule):

            silver.loc[
                mask,
                "quality_warnings"
            ] = (
                silver.loc[
                    mask,
                    "quality_warnings"
                ]
                .apply(
                    lambda x:
                    x + [rule]
                )
            )

        # ====================================================
        # 4. BLOCKING RULES
        # ====================================================

        add_error(
            silver["trip_id"].isna(),
            "MISSING_TRIP_ID"
        )

        add_error(
            silver[
                "trip_start_timestamp"
            ].isna(),
            "INVALID_START_TIMESTAMP"
        )

        add_error(
            silver[
                "trip_end_timestamp"
            ].isna(),
            "INVALID_END_TIMESTAMP"
        )

        add_error(

            (
                silver[
                    "trip_start_timestamp"
                ].notna()
                &
                silver[
                    "trip_end_timestamp"
                ].notna()
                &
                (
                    silver[
                        "trip_end_timestamp"
                    ]
                    <
                    silver[
                        "trip_start_timestamp"
                    ]
                )
            ),

            "END_BEFORE_START"
        )

        add_error(

            silver[
                "trip_seconds"
            ].notna()
            &
            (
                silver[
                    "trip_seconds"
                ] <= 0
            ),

            "INVALID_DURATION"
        )

        add_error(

            silver[
                "trip_miles"
            ].notna()
            &
            (
                silver[
                    "trip_miles"
                ] < 0
            ),

            "NEGATIVE_DISTANCE"
        )

        for col in [

            "fare",
            "tips",
            "tolls",
            "extras",
            "trip_total",

        ]:

            add_error(

                silver[col].notna()
                &
                (
                    silver[col] < 0
                ),

                f"NEGATIVE_{col.upper()}"
            )

        # ====================================================
        # 5. WARNING RULES
        # ====================================================

        add_warning(

            silver[
                "trip_miles"
            ].notna()
            &
            (
                silver[
                    "trip_miles"
                ] == 0
            ),

            "ZERO_DISTANCE"
        )

        silver[
            "calculated_trip_seconds"
        ] = (

            silver[
                "trip_end_timestamp"
            ]
            -
            silver[
                "trip_start_timestamp"
            ]

        ).dt.total_seconds()

        add_warning(

            (
                silver[
                    "trip_seconds"
                ].notna()
                &
                silver[
                    "calculated_trip_seconds"
                ].notna()
                &
                (

                    (
                        silver[
                            "trip_seconds"
                        ]
                        -
                        silver[
                            "calculated_trip_seconds"
                        ]
                    ).abs()

                    > 60

                )
            ),

            "DURATION_MISMATCH"
        )

        # Partial pickup coordinates

        add_warning(

            (
                silver[
                    "pickup_centroid_latitude"
                ].isna()
                ^
                silver[
                    "pickup_centroid_longitude"
                ].isna()
            ),

            "PARTIAL_PICKUP_COORDINATES"
        )

        # Partial dropoff coordinates

        add_warning(

            (
                silver[
                    "dropoff_centroid_latitude"
                ].isna()
                ^
                silver[
                    "dropoff_centroid_longitude"
                ].isna()
            ),

            "PARTIAL_DROPOFF_COORDINATES"
        )

        # ====================================================
        # 6. DUPLICATES
        # ====================================================

        duplicate_subset = [
            col
            for col in silver.columns
            if col not in [
                "quality_errors",
                "quality_warnings",
            ]
        ]

        silver[
            "is_exact_duplicate"
        ] = silver.duplicated(
            subset=duplicate_subset,
            keep=False
        )

        silver[
            "is_trip_id_duplicate"
        ] = (

            silver[
                "trip_id"
            ]
            .duplicated(
                keep=False
            )
        )

        add_warning(

            silver[
                "is_trip_id_duplicate"
            ],

            "DUPLICATED_TRIP_ID"
        )

        # ====================================================
        # 7. DERIVED COLUMNS
        # ====================================================

        silver[
            "trip_date"
        ] = (

            silver[
                "trip_start_timestamp"
            ]
            .dt.date
        )

        silver[
            "trip_year"
        ] = (

            silver[
                "trip_start_timestamp"
            ]
            .dt.year
        )

        silver[
            "trip_month"
        ] = (

            silver[
                "trip_start_timestamp"
            ]
            .dt.month
        )

        silver[
            "trip_hour"
        ] = (

            silver[
                "trip_start_timestamp"
            ]
            .dt.hour
        )

        silver[
            "trip_duration_minutes"
        ] = (

            silver[
                "trip_seconds"
            ]
            / 60
        )

        silver[
            "has_tip"
        ] = (

            silver[
                "tips"
            ]
            .fillna(0)
            > 0
        )

        silver[
            "tip_rate"
        ] = np.where(

            silver[
                "fare"
            ] > 0,

            silver[
                "tips"
            ]
            /
            silver[
                "fare"
            ],

            np.nan
        )

        # ====================================================
        # 8. ERROR / WARNING FLAGS
        # ====================================================

        silver[
            "has_error"
        ] = (

            silver[
                "quality_errors"
            ]
            .apply(
                lambda x:
                len(x) > 0
            )
        )

        silver[
            "has_warning"
        ] = (

            silver[
                "quality_warnings"
            ]
            .apply(
                lambda x:
                len(x) > 0
            )
        )

        # ====================================================
        # 9. SPLIT SILVER / QUARANTINE
        # ====================================================

        valid_df = (

            silver[
                ~silver[
                    "has_error"
                ]
            ]
            .copy()
        )

        rejected_df = (

            silver[
                silver[
                    "has_error"
                ]
            ]
            .copy()
        )

        # ====================================================
        # 10. SAVE DATASETS
        # ====================================================

        valid_path = (
            output_dir
            / "valid.csv"
        )

        rejected_path = (
            output_dir
            / "quarantine.csv"
        )

        valid_df.to_csv(
            valid_path,
            index=False
        )

        rejected_df.to_csv(
            rejected_path,
            index=False
        )

        # ====================================================
        # 11. COLUMN PROFILE
        # ====================================================

        profile = build_column_profile(
            silver
        )

        profile.to_csv(
            report_dir
            / "column_profile.csv",
            index=False
        )

        # ====================================================
        # 12. ERROR REPORT
        # ====================================================

        error_rows = []

        for rule in (
            silver[
                "quality_errors"
            ]
            .explode()
            .dropna()
            .unique()
        ):

            mask = (
                silver[
                    "quality_errors"
                ]
                .apply(
                    lambda x:
                    rule in x
                )
            )

            error_rows.append({
                "rule": rule,
                "count": int(
                    mask.sum()
                ),
                "rate": float(
                    mask.mean()
                )
            })

        error_report = pd.DataFrame(
            error_rows
        )

        error_report.to_csv(
            report_dir
            / "errors_report.csv",
            index=False
        )

        # ====================================================
        # 13. WARNING REPORT
        # ====================================================

        warning_rows = []

        for rule in (
            silver[
                "quality_warnings"
            ]
            .explode()
            .dropna()
            .unique()
        ):

            mask = (
                silver[
                    "quality_warnings"
                ]
                .apply(
                    lambda x:
                    rule in x
                )
            )

            warning_rows.append({
                "rule": rule,
                "count": int(
                    mask.sum()
                ),
                "rate": float(
                    mask.mean()
                )
            })

        warning_report = pd.DataFrame(
            warning_rows
        )

        warning_report.to_csv(
            report_dir
            / "warnings_report.csv",
            index=False
        )

        # ====================================================
        # 14. DQ SUMMARY
        # ====================================================

        dq_report = {

            "run_id":
                run_id,

            "source_run_id":
                bronze_run_id,

            "input_rows":
                len(silver),

            "valid_rows":
                len(valid_df),

            "rejected_rows":
                len(rejected_df),

            "warning_rows":
                int(
                    silver[
                        "has_warning"
                    ].sum()
                ),

            "rejection_rate":
                (
                    len(rejected_df)
                    /
                    len(silver)
                    if len(silver)
                    else 0
                ),

            "duplicate_trip_rows":
                int(
                    silver[
                        "is_trip_id_duplicate"
                    ]
                    .sum()
                ),

            "exact_duplicate_rows":
                int(
                    silver[
                        "is_exact_duplicate"
                    ]
                    .sum()
                ),

            "generated_at":
                utc_now()
        }

        save_json(
            dq_report,
            report_dir
            / "dq_report.json"
        )

        # ====================================================
        # 15. FINAL MANIFEST
        # ====================================================

        manifest.update({

            "status":
                "SUCCESS",

            "finished_at":
                utc_now(),

            "valid_rows":
                len(valid_df),

            "rejected_rows":
                len(rejected_df),

            "warning_rows":
                int(
                    silver[
                        "has_warning"
                    ]
                    .sum()
                ),

            "valid_path":
                str(valid_path),

            "quarantine_path":
                str(rejected_path),

        })

        save_json(
            manifest,
            report_dir
            / "manifest.json"
        )

        logger.info(
            f"Silver completed | "
            f"run_id={run_id} | "
            f"source_run_id={bronze_run_id} | "
            f"input={len(silver)} | "
            f"valid={len(valid_df)} | "
            f"rejected={len(rejected_df)}"
        )

        return {
            "silver_df": silver,
            "valid_df": valid_df,
            "rejected_df": rejected_df,
            "dq_report": dq_report,
            "errors_report": error_report,
            "warnings_report": warning_report,
            "manifest": manifest
        }

    except Exception as exc:

        manifest.update({

            "status":
                "FAILED",

            "finished_at":
                utc_now(),

            "error":
                str(exc)

        })

        save_json(
            manifest,
            report_dir
            / "manifest.json"
        )

        logger.exception(
            f"Silver failed | "
            f"run_id={run_id}"
        )

        raise
````

````python
def build_silver(
    bronze_df,
    bronze_run_id,
    run_id=None
):

    # ========================================================
    # INITIALIZATION
    # ========================================================

    run_id = run_id or generate_run_id(
        "SLV"
    )

    logger = create_logger(
        "silver",
        run_id
    )

    started_at = utc_now()

    output_dir = (
        SILVER_DIR
        / f"run_id={run_id}"
    )

    output_dir.mkdir(
        parents=True,
        exist_ok=True
    )

    report_dir = (
        REPORT_DIR
        / "silver"
        / run_id
    )

    report_dir.mkdir(
        parents=True,
        exist_ok=True
    )

    logger.info(
        f"Silver run started | "
        f"run_id={run_id} | "
        f"source_run_id={bronze_run_id} | "
        f"rows={len(bronze_df)}"
    )

    manifest = {

        "run_id": run_id,

        "source_run_id": bronze_run_id,

        "layer": "silver",

        "dataset": "chicago_taxi",

        "status": "RUNNING",

        "started_at": started_at,

        "finished_at": None,

        "input_rows": len(bronze_df),

        "valid_rows": None,

        "rejected_rows": None,

        "warning_rows": None,

        "error": None,
    }

    save_json(
        manifest,
        report_dir / "manifest.json"
    )

    try:

        silver = bronze_df.copy()

        # ====================================================
        # 1. COLUMN NORMALIZATION
        # ====================================================

        silver.columns = (
            silver.columns
            .str.strip()
            .str.lower()
            .str.replace(
                r"\s+",
                "_",
                regex=True
            )
        )

        # ====================================================
        # 2. TYPE CASTING
        # ====================================================

        date_cols = [
            "trip_start_timestamp",
            "trip_end_timestamp",
        ]

        for col in date_cols:

            silver[col] = pd.to_datetime(
                silver[col],
                errors="coerce"
            )

        numeric_cols = [

            "trip_seconds",
            "trip_miles",

            "fare",
            "tips",
            "tolls",
            "extras",
            "trip_total",

            "pickup_centroid_latitude",
            "pickup_centroid_longitude",

            "dropoff_centroid_latitude",
            "dropoff_centroid_longitude",
        ]

        for col in numeric_cols:

            silver[col] = pd.to_numeric(
                silver[col],
                errors="coerce"
            )

        for col in [

            "pickup_community_area",
            "dropoff_community_area",

        ]:

            silver[col] = (
                pd.to_numeric(
                    silver[col],
                    errors="coerce"
                )
                .astype("Int64")
            )

        # ====================================================
        # 3. ERROR / WARNING ARRAYS
        # ====================================================

        silver["quality_errors"] = [
            []
            for _ in range(len(silver))
        ]

        silver["quality_warnings"] = [
            []
            for _ in range(len(silver))
        ]

        def add_error(mask, rule):

            silver.loc[
                mask,
                "quality_errors"
            ] = (
                silver.loc[
                    mask,
                    "quality_errors"
                ]
                .apply(
                    lambda x:
                    x + [rule]
                )
            )

        def add_warning(mask, rule):

            silver.loc[
                mask,
                "quality_warnings"
            ] = (
                silver.loc[
                    mask,
                    "quality_warnings"
                ]
                .apply(
                    lambda x:
                    x + [rule]
                )
            )

        # ====================================================
        # 4. BLOCKING RULES
        # ====================================================

        add_error(
            silver["trip_id"].isna(),
            "MISSING_TRIP_ID"
        )

        add_error(
            silver[
                "trip_start_timestamp"
            ].isna(),
            "INVALID_START_TIMESTAMP"
        )

        add_error(
            silver[
                "trip_end_timestamp"
            ].isna(),
            "INVALID_END_TIMESTAMP"
        )

        add_error(

            (
                silver[
                    "trip_start_timestamp"
                ].notna()
                &
                silver[
                    "trip_end_timestamp"
                ].notna()
                &
                (
                    silver[
                        "trip_end_timestamp"
                    ]
                    <
                    silver[
                        "trip_start_timestamp"
                    ]
                )
            ),

            "END_BEFORE_START"
        )

        add_error(

            silver[
                "trip_seconds"
            ].notna()
            &
            (
                silver[
                    "trip_seconds"
                ] <= 0
            ),

            "INVALID_DURATION"
        )

        add_error(

            silver[
                "trip_miles"
            ].notna()
            &
            (
                silver[
                    "trip_miles"
                ] < 0
            ),

            "NEGATIVE_DISTANCE"
        )

        for col in [

            "fare",
            "tips",
            "tolls",
            "extras",
            "trip_total",

        ]:

            add_error(

                silver[col].notna()
                &
                (
                    silver[col] < 0
                ),

                f"NEGATIVE_{col.upper()}"
            )

        # ====================================================
        # 5. WARNING RULES
        # ====================================================

        add_warning(

            silver[
                "trip_miles"
            ].notna()
            &
            (
                silver[
                    "trip_miles"
                ] == 0
            ),

            "ZERO_DISTANCE"
        )

        silver[
            "calculated_trip_seconds"
        ] = (

            silver[
                "trip_end_timestamp"
            ]
            -
            silver[
                "trip_start_timestamp"
            ]

        ).dt.total_seconds()

        add_warning(

            (
                silver[
                    "trip_seconds"
                ].notna()
                &
                silver[
                    "calculated_trip_seconds"
                ].notna()
                &
                (

                    (
                        silver[
                            "trip_seconds"
                        ]
                        -
                        silver[
                            "calculated_trip_seconds"
                        ]
                    ).abs()

                    > 60

                )
            ),

            "DURATION_MISMATCH"
        )

        # Partial pickup coordinates

        add_warning(

            (
                silver[
                    "pickup_centroid_latitude"
                ].isna()
                ^
                silver[
                    "pickup_centroid_longitude"
                ].isna()
            ),

            "PARTIAL_PICKUP_COORDINATES"
        )

        # Partial dropoff coordinates

        add_warning(

            (
                silver[
                    "dropoff_centroid_latitude"
                ].isna()
                ^
                silver[
                    "dropoff_centroid_longitude"
                ].isna()
            ),

            "PARTIAL_DROPOFF_COORDINATES"
        )

        # ====================================================
        # 6. DUPLICATES
        # ====================================================

        duplicate_subset = [
            col
            for col in silver.columns
            if col not in [
                "quality_errors",
                "quality_warnings",
            ]
        ]

        silver[
            "is_exact_duplicate"
        ] = silver.duplicated(
            subset=duplicate_subset,
            keep=False
        )

        silver[
            "is_trip_id_duplicate"
        ] = (

            silver[
                "trip_id"
            ]
            .duplicated(
                keep=False
            )
        )

        add_warning(

            silver[
                "is_trip_id_duplicate"
            ],

            "DUPLICATED_TRIP_ID"
        )

        # ====================================================
        # 7. DERIVED COLUMNS
        # ====================================================

        silver[
            "trip_date"
        ] = (

            silver[
                "trip_start_timestamp"
            ]
            .dt.date
        )

        silver[
            "trip_year"
        ] = (

            silver[
                "trip_start_timestamp"
            ]
            .dt.year
        )

        silver[
            "trip_month"
        ] = (

            silver[
                "trip_start_timestamp"
            ]
            .dt.month
        )

        silver[
            "trip_hour"
        ] = (

            silver[
                "trip_start_timestamp"
            ]
            .dt.hour
        )

        silver[
            "trip_duration_minutes"
        ] = (

            silver[
                "trip_seconds"
            ]
            / 60
        )

        silver[
            "has_tip"
        ] = (

            silver[
                "tips"
            ]
            .fillna(0)
            > 0
        )

        silver[
            "tip_rate"
        ] = np.where(

            silver[
                "fare"
            ] > 0,

            silver[
                "tips"
            ]
            /
            silver[
                "fare"
            ],

            np.nan
        )

        # ====================================================
        # 8. ERROR / WARNING FLAGS
        # ====================================================

        silver[
            "has_error"
        ] = (

            silver[
                "quality_errors"
            ]
            .apply(
                lambda x:
                len(x) > 0
            )
        )

        silver[
            "has_warning"
        ] = (

            silver[
                "quality_warnings"
            ]
            .apply(
                lambda x:
                len(x) > 0
            )
        )

        # ====================================================
        # 9. SPLIT SILVER / QUARANTINE
        # ====================================================

        valid_df = (

            silver[
                ~silver[
                    "has_error"
                ]
            ]
            .copy()
        )

        rejected_df = (

            silver[
                silver[
                    "has_error"
                ]
            ]
            .copy()
        )

        # ====================================================
        # 10. SAVE DATASETS
        # ====================================================

        valid_path = (
            output_dir
            / "valid.csv"
        )

        rejected_path = (
            output_dir
            / "quarantine.csv"
        )

        valid_df.to_csv(
            valid_path,
            index=False
        )

        rejected_df.to_csv(
            rejected_path,
            index=False
        )

        # ====================================================
        # 11. COLUMN PROFILE
        # ====================================================

        profile = build_column_profile(
            silver
        )

        profile.to_csv(
            report_dir
            / "column_profile.csv",
            index=False
        )

        # ====================================================
        # 12. ERROR REPORT
        # ====================================================

        error_rows = []

        for rule in (
            silver[
                "quality_errors"
            ]
            .explode()
            .dropna()
            .unique()
        ):

            mask = (
                silver[
                    "quality_errors"
                ]
                .apply(
                    lambda x:
                    rule in x
                )
            )

            error_rows.append({
                "rule": rule,
                "count": int(
                    mask.sum()
                ),
                "rate": float(
                    mask.mean()
                )
            })

        error_report = pd.DataFrame(
            error_rows
        )

        error_report.to_csv(
            report_dir
            / "errors_report.csv",
            index=False
        )

        # ====================================================
        # 13. WARNING REPORT
        # ====================================================

        warning_rows = []

        for rule in (
            silver[
                "quality_warnings"
            ]
            .explode()
            .dropna()
            .unique()
        ):

            mask = (
                silver[
                    "quality_warnings"
                ]
                .apply(
                    lambda x:
                    rule in x
                )
            )

            warning_rows.append({
                "rule": rule,
                "count": int(
                    mask.sum()
                ),
                "rate": float(
                    mask.mean()
                )
            })

        warning_report = pd.DataFrame(
            warning_rows
        )

        warning_report.to_csv(
            report_dir
            / "warnings_report.csv",
            index=False
        )

        # ====================================================
        # 14. DQ SUMMARY
        # ====================================================

        dq_report = {

            "run_id":
                run_id,

            "source_run_id":
                bronze_run_id,

            "input_rows":
                len(silver),

            "valid_rows":
                len(valid_df),

            "rejected_rows":
                len(rejected_df),

            "warning_rows":
                int(
                    silver[
                        "has_warning"
                    ].sum()
                ),

            "rejection_rate":
                (
                    len(rejected_df)
                    /
                    len(silver)
                    if len(silver)
                    else 0
                ),

            "duplicate_trip_rows":
                int(
                    silver[
                        "is_trip_id_duplicate"
                    ]
                    .sum()
                ),

            "exact_duplicate_rows":
                int(
                    silver[
                        "is_exact_duplicate"
                    ]
                    .sum()
                ),

            "generated_at":
                utc_now()
        }

        save_json(
            dq_report,
            report_dir
            / "dq_report.json"
        )

        # ====================================================
        # 15. FINAL MANIFEST
        # ====================================================

        manifest.update({

            "status":
                "SUCCESS",

            "finished_at":
                utc_now(),

            "valid_rows":
                len(valid_df),

            "rejected_rows":
                len(rejected_df),

            "warning_rows":
                int(
                    silver[
                        "has_warning"
                    ]
                    .sum()
                ),

            "valid_path":
                str(valid_path),

            "quarantine_path":
                str(rejected_path),

        })

        save_json(
            manifest,
            report_dir
            / "manifest.json"
        )

        logger.info(
            f"Silver completed | "
            f"run_id={run_id} | "
            f"source_run_id={bronze_run_id} | "
            f"input={len(silver)} | "
            f"valid={len(valid_df)} | "
            f"rejected={len(rejected_df)}"
        )

        return {
            "silver_df": silver,
            "valid_df": valid_df,
            "rejected_df": rejected_df,
            "dq_report": dq_report,
            "errors_report": error_report,
            "warnings_report": warning_report,
            "manifest": manifest
        }

    except Exception as exc:

        manifest.update({

            "status":
                "FAILED",

            "finished_at":
                utc_now(),

            "error":
                str(exc)

        })

        save_json(
            manifest,
            report_dir
            / "manifest.json"
        )

        logger.exception(
            f"Silver failed | "
            f"run_id={run_id}"
        )

        raise
````

````python
import os
import time
from html import escape

import matplotlib.pyplot as plt
import folium
from folium.plugins import HeatMap, MarkerCluster


GOLD_DIR = BASE_DIR / "gold" / "chicago_taxi"
MAP_DIR = BASE_DIR / "maps"
CACHE_DIR = BASE_DIR / "cache"
FOURSQUARE_CACHE_PATH = CACHE_DIR / "foursquare_places.json"


for path in [
    GOLD_DIR,
    REPORT_DIR / "gold",
    MAP_DIR,
    CACHE_DIR,
]:
    path.mkdir(parents=True, exist_ok=True)


def _safe_scalar(value):
    if pd.isna(value):
        return "N/A"
    return value


def _safe_float(value):
    if pd.isna(value):
        return None
    return float(value)


def _json_safe_dict(data):
    return json.loads(
        json.dumps(
            data,
            default=str,
            allow_nan=False,
        )
    )


def _load_json(path, default):
    path = Path(path)
    if not path.exists():
        return default

    try:
        with open(path, "r", encoding="utf-8") as f:
            return json.load(f)
    except json.JSONDecodeError:
        return default


def _save_json(data, path):
    save_json(
        _json_safe_dict(data),
        path,
    )


def _prepare_gold_frame(valid_df):
    work = valid_df.copy()

    required_columns = [
        "trip_id",
        "taxi_id",
        "trip_start_timestamp",
        "trip_end_timestamp",
        "trip_miles",
        "fare",
        "tips",
        "trip_total",
        "payment_type",
        "company",
        "pickup_community_area",
        "dropoff_community_area",
        "pickup_centroid_latitude",
        "pickup_centroid_longitude",
        "dropoff_centroid_latitude",
        "dropoff_centroid_longitude",
        "trip_date",
        "trip_hour",
        "trip_duration_minutes",
        "has_tip",
    ]

    missing_columns = [
        column
        for column in required_columns
        if column not in work.columns
    ]

    for column in missing_columns:
        work[column] = pd.NA

    for column in [
        "trip_miles",
        "fare",
        "tips",
        "trip_total",
        "trip_duration_minutes",
        "pickup_centroid_latitude",
        "pickup_centroid_longitude",
        "dropoff_centroid_latitude",
        "dropoff_centroid_longitude",
    ]:
        work[column] = pd.to_numeric(
            work[column],
            errors="coerce",
        )

    if work["has_tip"].isna().all() and "tips" in work.columns:
        work["has_tip"] = work["tips"].fillna(0) > 0
    else:
        work["has_tip"] = work["has_tip"].fillna(False).astype(bool)

    if "trip_id" in valid_df.columns:
        work["_trip_count"] = work["trip_id"].notna().astype(int)
    else:
        work["_trip_count"] = 1

    return work, missing_columns


def _add_tipped_rate(metrics):
    if "tipped_trip_count" in metrics.columns:
        metrics["tipped_trip_rate"] = np.where(
            metrics["trip_count"] > 0,
            metrics["tipped_trip_count"] / metrics["trip_count"],
            0,
        )
    return metrics


def build_daily_metrics(gold_df):
    daily = (
        gold_df
        .dropna(subset=["trip_date"])
        .groupby("trip_date", as_index=False)
        .agg(
            trip_count=("_trip_count", "sum"),
            total_revenue=("trip_total", "sum"),
            avg_revenue_per_trip=("trip_total", "mean"),
            avg_trip_duration_minutes=("trip_duration_minutes", "mean"),
            avg_trip_miles=("trip_miles", "mean"),
            total_trip_miles=("trip_miles", "sum"),
            total_tips=("tips", "sum"),
            avg_tip=("tips", "mean"),
            tipped_trip_count=("has_tip", "sum"),
        )
        .sort_values("trip_date")
        .reset_index(drop=True)
    )

    return _add_tipped_rate(daily)


def build_hourly_metrics(gold_df):
    hourly = (
        gold_df
        .dropna(subset=["trip_date", "trip_hour"])
        .groupby(["trip_date", "trip_hour"], as_index=False)
        .agg(
            trip_count=("_trip_count", "sum"),
            total_revenue=("trip_total", "sum"),
            avg_trip_total=("trip_total", "mean"),
            avg_trip_duration_minutes=("trip_duration_minutes", "mean"),
            avg_trip_miles=("trip_miles", "mean"),
        )
        .sort_values(["trip_date", "trip_hour"])
        .reset_index(drop=True)
    )

    return hourly


def build_pickup_zone_metrics(gold_df):
    zone_df = gold_df.copy()
    zone_df["pickup_community_area"] = (
        zone_df["pickup_community_area"]
        .astype("object")
        .where(zone_df["pickup_community_area"].notna(), "UNKNOWN")
    )

    zone = (
        zone_df
        .groupby("pickup_community_area", as_index=False, dropna=False)
        .agg(
            trip_count=("_trip_count", "sum"),
            total_revenue=("trip_total", "sum"),
            avg_trip_total=("trip_total", "mean"),
            avg_trip_duration_minutes=("trip_duration_minutes", "mean"),
            avg_trip_miles=("trip_miles", "mean"),
            total_tips=("tips", "sum"),
            tipped_trip_count=("has_tip", "sum"),
        )
        .sort_values("trip_count", ascending=False)
        .reset_index(drop=True)
    )

    zone = _add_tipped_rate(zone)
    return zone.drop(columns=["tipped_trip_count"])


def build_payment_metrics(gold_df):
    payment_df = gold_df.copy()
    payment_df["payment_type"] = (
        payment_df["payment_type"]
        .astype("object")
        .where(payment_df["payment_type"].notna(), "UNKNOWN")
    )

    payment = (
        payment_df
        .groupby("payment_type", as_index=False, dropna=False)
        .agg(
            trip_count=("_trip_count", "sum"),
            total_revenue=("trip_total", "sum"),
            avg_trip_total=("trip_total", "mean"),
            total_tips=("tips", "sum"),
            avg_tip=("tips", "mean"),
            avg_trip_duration_minutes=("trip_duration_minutes", "mean"),
            avg_trip_miles=("trip_miles", "mean"),
        )
        .sort_values("trip_count", ascending=False)
        .reset_index(drop=True)
    )

    return payment


def build_company_metrics(gold_df):
    company_df = gold_df.copy()
    company_df["company"] = (
        company_df["company"]
        .astype("object")
        .where(company_df["company"].notna(), "UNKNOWN")
    )

    company = (
        company_df
        .groupby("company", as_index=False, dropna=False)
        .agg(
            trip_count=("_trip_count", "sum"),
            total_revenue=("trip_total", "sum"),
            avg_trip_total=("trip_total", "mean"),
            avg_trip_duration_minutes=("trip_duration_minutes", "mean"),
            avg_trip_miles=("trip_miles", "mean"),
            total_tips=("tips", "sum"),
            tipped_trip_count=("has_tip", "sum"),
        )
        .sort_values("trip_count", ascending=False)
        .reset_index(drop=True)
    )

    company = _add_tipped_rate(company)
    return company.drop(columns=["tipped_trip_count"])


def build_kpi_summary(gold_df):
    total_trips = int(gold_df["_trip_count"].sum())
    tipped_trips = int(gold_df["has_tip"].sum())

    return {
        "total_trips": total_trips,
        "total_revenue": _safe_float(gold_df["trip_total"].sum()),
        "avg_trip_total": _safe_float(gold_df["trip_total"].mean()),
        "avg_trip_miles": _safe_float(gold_df["trip_miles"].mean()),
        "avg_trip_duration_minutes": _safe_float(
            gold_df["trip_duration_minutes"].mean()
        ),
        "total_tips": _safe_float(gold_df["tips"].sum()),
        "tipped_trip_rate": (
            tipped_trips / total_trips
            if total_trips
            else 0
        ),
        "unique_taxis": (
            int(gold_df["taxi_id"].nunique(dropna=True))
            if "taxi_id" in gold_df.columns
            else None
        ),
        "unique_companies": (
            int(gold_df["company"].nunique(dropna=True))
            if "company" in gold_df.columns
            else None
        ),
    }


def build_validation_report(
    gold_df,
    daily_metrics,
    payment_metrics,
    company_metrics,
    pickup_zone_metrics,
):
    input_rows = int(len(gold_df))

    checks = {
        "daily": int(daily_metrics["trip_count"].sum()),
        "payment": int(payment_metrics["trip_count"].sum()),
        "company": int(company_metrics["trip_count"].sum()),
        "zone": int(pickup_zone_metrics["trip_count"].sum()),
    }

    return {
        "input_rows": input_rows,
        "daily_trip_count_matches": checks["daily"] == input_rows,
        "daily_difference": checks["daily"] - input_rows,
        "payment_trip_count_matches": checks["payment"] == input_rows,
        "payment_difference": checks["payment"] - input_rows,
        "company_trip_count_matches": checks["company"] == input_rows,
        "company_difference": checks["company"] - input_rows,
        "zone_trip_count_matches": checks["zone"] == input_rows,
        "zone_difference": checks["zone"] - input_rows,
        "unknown_policy": {
            "payment_type": "Null values grouped as UNKNOWN",
            "company": "Null values grouped as UNKNOWN",
            "pickup_community_area": "Null values grouped as UNKNOWN",
        },
    }


def _foursquare_cache_key(lat, lon, radius=300, limit=5):
    return f"{round(float(lat), 5)}:{round(float(lon), 5)}:{radius}:{limit}"


def _parse_foursquare_place(place):
    categories = place.get("categories") or []
    location = place.get("location") or {}

    return {
        "place_name": place.get("name"),
        "category": (
            categories[0].get("name")
            if categories
            else None
        ),
        "distance": place.get("distance"),
        "address": (
            location.get("formatted_address")
            or location.get("address")
        ),
    }


def get_nearby_places(
    lat,
    lon,
    radius=300,
    limit=5,
    api_key=None,
    cache=None,
    stats=None,
    logger=None,
    timeout=10,
):
    cache = cache if cache is not None else {}
    stats = stats if stats is not None else {
        "calls": 0,
        "cache_hits": 0,
        "errors": 0,
    }

    if not api_key:
        return []

    cache_key = _foursquare_cache_key(
        lat,
        lon,
        radius=radius,
        limit=limit,
    )

    if cache_key in cache:
        stats["cache_hits"] += 1
        return cache[cache_key]

    try:
        stats["calls"] += 1

        response = requests.get(
            "https://api.foursquare.com/v3/places/search",
            headers={
                "Accept": "application/json",
                "Authorization": api_key,
            },
            params={
                "ll": f"{float(lat)},{float(lon)}",
                "radius": radius,
                "limit": limit,
            },
            timeout=timeout,
        )

        response.raise_for_status()
        payload = response.json()
        places = [
            _parse_foursquare_place(place)
            for place in payload.get("results", [])
        ]

        cache[cache_key] = places
        return places

    except Exception as exc:
        stats["errors"] += 1
        cache[cache_key] = []

        if logger:
            logger.warning(
                "Foursquare enrichment failed for %s,%s: %s",
                lat,
                lon,
                exc,
            )

        return []


def _format_places_for_popup(places, max_places=5):
    if not places:
        return "N/A"

    lines = []
    for place in places[:max_places]:
        name = escape(str(place.get("place_name") or "Unknown place"))
        category = escape(str(place.get("category") or "Unknown category"))
        distance = place.get("distance")
        distance_label = (
            f" - {int(distance)} m"
            if distance is not None
            else ""
        )
        lines.append(f"<li>{name} ({category}{distance_label})</li>")

    return "<ul>" + "".join(lines) + "</ul>"


def _build_trip_popup(row, nearby_places=None):
    fields = [
        ("Trip ID", row.get("trip_id")),
        ("Taxi ID", row.get("taxi_id")),
        ("Start Timestamp", row.get("trip_start_timestamp")),
        ("End Timestamp", row.get("trip_end_timestamp")),
        ("Duration", row.get("trip_duration_minutes")),
        ("Distance", row.get("trip_miles")),
        ("Fare", row.get("fare")),
        ("Tips", row.get("tips")),
        ("Trip Total", row.get("trip_total")),
        ("Payment Type", row.get("payment_type")),
        ("Company", row.get("company")),
        ("Pickup Community Area", row.get("pickup_community_area")),
        ("Dropoff Community Area", row.get("dropoff_community_area")),
    ]

    html_lines = [
        "<b>Historical trip</b>",
        "<br><i>Line is pickup to dropoff only, not a GPS route.</i>",
        "<hr>",
    ]

    for label, value in fields:
        html_lines.append(
            f"<b>{escape(label)}:</b> {escape(str(_safe_scalar(value)))}<br>"
        )

    html_lines.append("<hr><b>Nearby Places</b>")
    html_lines.append(_format_places_for_popup(nearby_places or []))

    return "".join(html_lines)


def _build_geo_view(gold_df):
    geo_columns = [
        "pickup_centroid_latitude",
        "pickup_centroid_longitude",
        "dropoff_centroid_latitude",
        "dropoff_centroid_longitude",
    ]

    return (
        gold_df
        .dropna(subset=geo_columns)
        .copy()
    )


def _sample_map_trips(geo_df, map_sample_size=500):
    sample_size = min(
        int(map_sample_size),
        len(geo_df),
    )

    if sample_size <= 0:
        return geo_df.head(0).copy()

    return (
        geo_df
        .sample(
            n=sample_size,
            random_state=42,
        )
        .copy()
    )


def _build_folium_map(
    map_sample_df,
    run_id,
    enable_foursquare=True,
    logger=None,
):
    fmap = folium.Map(
        location=[41.8781, -87.6298],
        zoom_start=11,
        tiles=None,
    )

    carto_key = os.getenv("CARTO_BASEMAPS_API_KEY")

    if carto_key:
        folium.TileLayer(
            tiles=(
                "https://basemaps.cartocdn.com/rastertiles/"
                "voyager/{z}/{x}/{y}.png?key="
                f"{carto_key}"
            ),
            attr=(
                '&copy; <a href="https://www.openstreetmap.org/copyright">'
                'OpenStreetMap</a> contributors &copy; '
                '<a href="https://carto.com/attributions">CARTO</a>'
            ),
            name="CARTO Voyager",
            control=True,
        ).add_to(fmap)
    else:
        folium.TileLayer(
            tiles="OpenStreetMap",
            name="OpenStreetMap fallback",
            control=True,
        ).add_to(fmap)


    pickups_group = folium.FeatureGroup(
        name="Pickups",
        show=True,
    )
    dropoffs_group = folium.FeatureGroup(
        name="Dropoffs",
        show=True,
    )
    lines_group = folium.FeatureGroup(
        name="Trip Lines",
        show=True,
    )
    heatmap_group = folium.FeatureGroup(
        name="Pickup HeatMap",
        show=False,
    )

    pickup_cluster = MarkerCluster(name="Pickup clusters")
    dropoff_cluster = MarkerCluster(name="Dropoff clusters")

    foursquare_stats = {
        "enabled": bool(enable_foursquare),
        "skipped": False,
        "calls": 0,
        "cache_hits": 0,
        "errors": 0,
        "cache_path": str(FOURSQUARE_CACHE_PATH),
    }

    api_key = os.getenv("FOURSQUARE_API_KEY")
    cache = _load_json(FOURSQUARE_CACHE_PATH, {})
    nearby_by_key = {}

    if enable_foursquare and not api_key:
        foursquare_stats["skipped"] = True
        if logger:
            logger.warning(
                "FOURSQUARE_API_KEY not set; skipping Foursquare enrichment"
            )

    if enable_foursquare and api_key and not map_sample_df.empty:
        unique_pickups = (
            map_sample_df[
                [
                    "pickup_centroid_latitude",
                    "pickup_centroid_longitude",
                ]
            ]
            .drop_duplicates()
        )

        for _, point in unique_pickups.iterrows():
            lat = point["pickup_centroid_latitude"]
            lon = point["pickup_centroid_longitude"]
            key = _foursquare_cache_key(lat, lon)
            nearby_by_key[key] = get_nearby_places(
                lat,
                lon,
                api_key=api_key,
                cache=cache,
                stats=foursquare_stats,
                logger=logger,
            )

        _save_json(cache, FOURSQUARE_CACHE_PATH)

    pickup_heat_points = []

    for _, row in map_sample_df.iterrows():
        pickup = [
            float(row["pickup_centroid_latitude"]),
            float(row["pickup_centroid_longitude"]),
        ]
        dropoff = [
            float(row["dropoff_centroid_latitude"]),
            float(row["dropoff_centroid_longitude"]),
        ]

        pickup_key = _foursquare_cache_key(
            pickup[0],
            pickup[1],
        )
        nearby_places = nearby_by_key.get(pickup_key, [])
        popup = folium.Popup(
            _build_trip_popup(
                row,
                nearby_places=nearby_places,
            ),
            max_width=420,
        )

        folium.Marker(
            location=pickup,
            popup=popup,
            tooltip="Pickup",
            icon=folium.Icon(
                color="green",
                icon="play",
            ),
        ).add_to(pickup_cluster)

        folium.Marker(
            location=dropoff,
            popup=popup,
            tooltip="Dropoff",
            icon=folium.Icon(
                color="red",
                icon="stop",
            ),
        ).add_to(dropoff_cluster)

        folium.PolyLine(
            locations=[pickup, dropoff],
            color="#3366cc",
            weight=2,
            opacity=0.45,
            tooltip="Pickup to dropoff link, not GPS route",
        ).add_to(lines_group)

        pickup_heat_points.append(pickup)

    pickup_cluster.add_to(pickups_group)
    dropoff_cluster.add_to(dropoffs_group)

    if pickup_heat_points:
        HeatMap(
            pickup_heat_points,
            radius=14,
            blur=18,
            name="Pickup density",
        ).add_to(heatmap_group)

    pickups_group.add_to(fmap)
    dropoffs_group.add_to(fmap)
    lines_group.add_to(fmap)
    heatmap_group.add_to(fmap)

    folium.LayerControl().add_to(fmap)

    return fmap, foursquare_stats


def build_gold(
    valid_df,
    silver_run_id,
    map_sample_size=500,
    enable_foursquare=True,
):
    run_id = generate_run_id("GLD")
    started_at = utc_now()
    start_time = time.time()

    output_dir = GOLD_DIR / f"run_id={run_id}"
    report_dir = REPORT_DIR / "gold" / run_id
    map_dir = MAP_DIR / run_id

    for path in [
        output_dir,
        report_dir,
        map_dir,
    ]:
        path.mkdir(parents=True, exist_ok=True)

    logger = create_logger(
        "gold",
        run_id,
    )

    manifest = {
        "run_id": run_id,
        "source_run_id": silver_run_id,
        "layer": "gold",
        "dataset": "chicago_taxi",
        "status": "RUNNING",
        "started_at": started_at,
        "finished_at": None,
        "input_rows": int(len(valid_df)),
        "outputs": {},
        "error": None,
    }

    _save_json(
        manifest,
        report_dir / "manifest.json",
    )

    try:
        logger.info(
            "Gold run started | run_id=%s | source_run_id=%s | rows=%s",
            run_id,
            silver_run_id,
            len(valid_df),
        )

        gold_df, missing_columns = _prepare_gold_frame(valid_df)

        daily_metrics = build_daily_metrics(gold_df)
        hourly_metrics = build_hourly_metrics(gold_df)
        pickup_zone_metrics = build_pickup_zone_metrics(gold_df)
        payment_metrics = build_payment_metrics(gold_df)
        company_metrics = build_company_metrics(gold_df)

        kpi_summary = build_kpi_summary(gold_df)
        validation_report = build_validation_report(
            gold_df,
            daily_metrics,
            payment_metrics,
            company_metrics,
            pickup_zone_metrics,
        )
        validation_report["missing_columns"] = missing_columns

        output_paths = {
            "daily_metrics": output_dir / "daily_metrics.csv",
            "hourly_metrics": output_dir / "hourly_metrics.csv",
            "pickup_zone_metrics": output_dir / "pickup_zone_metrics.csv",
            "payment_metrics": output_dir / "payment_metrics.csv",
            "company_metrics": output_dir / "company_metrics.csv",
        }

        daily_metrics.to_csv(
            output_paths["daily_metrics"],
            index=False,
        )
        hourly_metrics.to_csv(
            output_paths["hourly_metrics"],
            index=False,
        )
        pickup_zone_metrics.to_csv(
            output_paths["pickup_zone_metrics"],
            index=False,
        )
        payment_metrics.to_csv(
            output_paths["payment_metrics"],
            index=False,
        )
        company_metrics.to_csv(
            output_paths["company_metrics"],
            index=False,
        )

        geo_df = _build_geo_view(gold_df)
        map_sample_df = _sample_map_trips(
            geo_df,
            map_sample_size=map_sample_size,
        )

        fmap, foursquare_stats = _build_folium_map(
            map_sample_df,
            run_id=run_id,
            enable_foursquare=enable_foursquare,
            logger=logger,
        )

        map_path = map_dir / "trips_map.html"
        fmap.save(str(map_path))

        gold_report = {
            "run_id": run_id,
            "source_run_id": silver_run_id,
            "kpi_summary": kpi_summary,
            "gold_rows": {
                "daily_metrics": int(len(daily_metrics)),
                "hourly_metrics": int(len(hourly_metrics)),
                "pickup_zone_metrics": int(len(pickup_zone_metrics)),
                "payment_metrics": int(len(payment_metrics)),
                "company_metrics": int(len(company_metrics)),
                "geo_trips": int(len(geo_df)),
                "map_sample": int(len(map_sample_df)),
            },
            "foursquare": foursquare_stats,
            "generated_at": utc_now(),
        }

        _save_json(
            gold_report,
            report_dir / "gold_report.json",
        )
        _save_json(
            validation_report,
            report_dir / "validation_report.json",
        )

        manifest.update({
            "status": "SUCCESS",
            "finished_at": utc_now(),
            "outputs": {
                **{
                    key: str(path)
                    for key, path in output_paths.items()
                },
                "gold_report": str(report_dir / "gold_report.json"),
                "validation_report": str(report_dir / "validation_report.json"),
                "map": str(map_path),
            },
        })

        _save_json(
            manifest,
            report_dir / "manifest.json",
        )

        duration_seconds = round(
            time.time() - start_time,
            3,
        )

        logger.info(
            "Gold completed | run_id=%s | source_run_id=%s | input_rows=%s | "
            "daily_rows=%s | hourly_rows=%s | zone_rows=%s | payment_rows=%s | "
            "company_rows=%s | map_sample_size=%s | foursquare_calls=%s | "
            "foursquare_cache_hits=%s | status=%s | duration_seconds=%s",
            run_id,
            silver_run_id,
            len(valid_df),
            len(daily_metrics),
            len(hourly_metrics),
            len(pickup_zone_metrics),
            len(payment_metrics),
            len(company_metrics),
            len(map_sample_df),
            foursquare_stats["calls"],
            foursquare_stats["cache_hits"],
            manifest["status"],
            duration_seconds,
        )

        return {
            "daily_metrics": daily_metrics,
            "hourly_metrics": hourly_metrics,
            "pickup_zone_metrics": pickup_zone_metrics,
            "payment_metrics": payment_metrics,
            "company_metrics": company_metrics,
            "kpi_summary": kpi_summary,
            "validation_report": validation_report,
            "gold_report": gold_report,
            "geo_df": geo_df,
            "map_sample": map_sample_df,
            "map": fmap,
            "foursquare_stats": foursquare_stats,
            "manifest": manifest,
        }

    except Exception as exc:
        manifest.update({
            "status": "FAILED",
            "finished_at": utc_now(),
            "error": str(exc),
        })

        _save_json(
            manifest,
            report_dir / "manifest.json",
        )

        logger.exception(
            "Gold failed | run_id=%s",
            run_id,
        )

        raise
````

````python
def plot_gold_bi(gold_result):
    daily_metrics = gold_result["daily_metrics"]
    hourly_metrics = gold_result["hourly_metrics"]
    pickup_zone_metrics = gold_result["pickup_zone_metrics"]
    payment_metrics = gold_result["payment_metrics"]
    company_metrics = gold_result["company_metrics"]

    plt.figure(figsize=(10, 4))
    daily_metrics.plot(
        x="trip_date",
        y="trip_count",
        kind="line",
        marker="o",
        ax=plt.gca(),
        legend=False,
        title="Daily Trip Volume",
    )
    plt.xlabel("Trip date")
    plt.ylabel("Trips")
    plt.tight_layout()
    plt.show()

    plt.figure(figsize=(10, 4))
    daily_metrics.plot(
        x="trip_date",
        y="total_revenue",
        kind="bar",
        ax=plt.gca(),
        legend=False,
        title="Daily Revenue",
    )
    plt.xlabel("Trip date")
    plt.ylabel("Revenue")
    plt.tight_layout()
    plt.show()

    trips_by_hour = (
        hourly_metrics
        .groupby("trip_hour", as_index=False)["trip_count"]
        .sum()
        .sort_values("trip_hour")
    )

    plt.figure(figsize=(10, 4))
    trips_by_hour.plot(
        x="trip_hour",
        y="trip_count",
        kind="bar",
        ax=plt.gca(),
        legend=False,
        title="Trips by Hour",
    )
    plt.xlabel("Hour")
    plt.ylabel("Trips")
    plt.tight_layout()
    plt.show()

    top_zones = pickup_zone_metrics.head(15)

    plt.figure(figsize=(10, 5))
    top_zones.plot(
        x="pickup_community_area",
        y="total_revenue",
        kind="bar",
        ax=plt.gca(),
        legend=False,
        title="Revenue by Pickup Community Area",
    )
    plt.xlabel("Pickup community area")
    plt.ylabel("Revenue")
    plt.tight_layout()
    plt.show()

    plt.figure(figsize=(10, 4))
    payment_metrics.plot(
        x="payment_type",
        y="trip_count",
        kind="bar",
        ax=plt.gca(),
        legend=False,
        title="Trips by Payment Type",
    )
    plt.xlabel("Payment type")
    plt.ylabel("Trips")
    plt.tight_layout()
    plt.show()

    top_companies = company_metrics.head(15)

    plt.figure(figsize=(10, 5))
    top_companies.plot(
        x="company",
        y="trip_count",
        kind="bar",
        ax=plt.gca(),
        legend=False,
        title="Trips by Company",
    )
    plt.xlabel("Company")
    plt.ylabel("Trips")
    plt.tight_layout()
    plt.show()


plot_gold_bi(gold_result)
````
